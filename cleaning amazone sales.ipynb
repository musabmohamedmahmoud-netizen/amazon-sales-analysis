```python
my_dict = {
    'Male': 'Male',
    'Female': 'Female',
    'male': 'Male',
    'M': 'Male',
    'F': 'Female',
    'female': 'Female',
    'FEMALE': 'Female',
    'MALE': 'Male'
}
```

```python
rgb['Gender'] = rgb['Gender'].map(my_dict)
```

```python
rgb['Gender'].value_counts()
```

```python
rgb["Age"]
```

```python
rgb[rgb['Age'] < 10]
```

```python
rgb[rgb['Age'] > 100]
```

```python
valid_mean = rgb.loc[(rgb['Age'] >= 10) & (rgb['Age'] <= 100), 'Age'].mean()
```

```python
rgb['Age'] = np.where((rgb['Age'] < 10) | (rgb['Age'] > 100), valid_mean, rgb['Age'])
```

```python
from ast import Num
rgb['Profit_Margin'] = ((rgb['Net_Profit_USD'] / rgb['Total_Sales_USD']) * 100).round(2).astype(str) + '%'
```

```python
rgb['Profit_Margin_Pct'] = ((rgb['Net_Profit_USD'] / rgb['Total_Sales_USD']) * 100).round(2)
```

```python
rgb[['Total_Sales_USD', 'Total_Cost_USD', 'Net_Profit_USD', 'Profit_Margin', 'Profit_Margin_Pct']].head()
```

```python
import numpy as np

conditions = [
    rgb['Profit_Margin_Pct'] > 50,
    (rgb['Profit_Margin_Pct'] >= 30) & (rgb['Profit_Margin_Pct'] <= 50),
    rgb['Profit_Margin_Pct'] < 30
]

choices = ['High Value', 'Medium Value', 'Low Value']

rgb['Customer_Category'] = np.select(conditions, choices, default='Medium Value')

rgb[['Profit_Margin_Pct', 'Customer_Category']].head()
```

```python
rgb
```

```python
rgb['Customer_Category'].value_counts()
```

```python
rgb['Order_Date'] = pd.to_datetime(rgb['Order_Date'])

rgb['Day_Name'] = rgb['Order_Date'].dt.day_name()

rgb['Month_Name'] = rgb['Order_Date'].dt.month_name()

rgb['Month_Num'] = rgb['Order_Date'].dt.month
```

```python
rgb
```

```python
rgb['Payment_Method'].value_counts().plot(kind='bar')
plt.title('Number of Transactions by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Count')
plt.show()
```

```python
rgb['Product_Category'].value_counts().plot(kind='bar')
plt.title('Number of category')
plt.xlabel('category')
plt.ylabel('Count')
plt.show()
```

```python
rgb['Gender'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('Gender')
plt.title('Gender')
plt.title('gender')
plt.xlabel('gender')
plt.ylabel('Count')
plt.show()
```
